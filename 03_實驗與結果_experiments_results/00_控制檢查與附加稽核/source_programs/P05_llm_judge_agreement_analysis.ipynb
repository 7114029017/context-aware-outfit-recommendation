# 清單概念一致性

```python
import json
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator, FormatStrFormatter
from scipy.spatial.distance import jensenshannon
from sentence_transformers import SentenceTransformer

# Jupyter / non-Jupyter 兼容（避免純 python 跑起來 display() 爆掉）
try:
    from IPython.display import display
except Exception:
    def display(x):
        print(x)

# =============================
# Config
# =============================
path_a = Path("Gemma3") / "checklist_C_star_Gemma3.json"
path_b = Path("Qwen3_Instruct") / "checklist_C_star_Qwen3VL32B.json"

OUT_DIR = Path("outputs")
OUT_DIR.mkdir(parents=True, exist_ok=True)

OUT_PREFIX = "m2m_nomicv2_rubric_similarity"

EMB_MODEL = "nomic-ai/nomic-embed-text-v2-moe"
PROMPT_A = "passage"
PROMPT_B = "passage"

# ✅ 重點：只看 0.5 -> 1.0（單張圖就夠清楚）
PLOT_TAU_MIN = 0.5
PLOT_TAU_MAX = 1.0
PLOT_TAU_STEP = 0.01   # 想要每 0.05 一格就改 0.05

# x 軸刻度（直覺好讀）
X_MAJOR_TICK = 0.05
X_MINOR_TICK = 0.01

# ✅ focus tau 仍然可用分位數自動抓（用來列 uncovered）
AUTO_TAU_FOCUS = True
TAU_FOCUS_Q = 0.90       # 用 maxSim 的 90% 分位當 focus
TAU_FOCUS = 0.60         # 若 AUTO_TAU_FOCUS=False 才用這個

SAVE_FIG = True
IMPORTANT_ONLY = True     # ✅ 只存重要檔案（避免存太多）

ENCODE_BATCH_SIZE = 64

# =============================
# Helpers
# =============================
def outpath(filename: str) -> Path:
    return OUT_DIR / filename

def save_csv(df: pd.DataFrame, filename: str, enabled: bool = True):
    if enabled:
        df.to_csv(outpath(filename), index=False, encoding="utf-8-sig")

def save_fig(filename: str, enabled: bool = True, dpi: int = 200):
    if enabled:
        plt.savefig(outpath(filename), dpi=dpi)

def load_checklist(path: Path) -> pd.DataFrame:
    if not path.exists():
        raise FileNotFoundError(f"找不到檔案：{path.resolve()}")

    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)

    if not (isinstance(data, dict) and "checklist" in data and isinstance(data["checklist"], list)):
        raise ValueError("JSON 格式不符：預期 dict 且包含 checklist(list)")

    df = pd.DataFrame(data["checklist"]).copy()
    required_cols = {"id", "text", "category", "weight"}
    missing = required_cols - set(df.columns)
    if missing:
        raise ValueError(f"JSON checklist 缺少欄位：{missing}")

    df["id"] = df["id"].astype(str)
    df["text"] = df["text"].astype(str).str.strip()
    df["category"] = df["category"].astype(str)
    df["weight"] = df["weight"].astype(float)
    return df[["id", "text", "category", "weight"]]

def category_distribution(df: pd.DataFrame) -> pd.Series:
    s = df.groupby("category")["weight"].sum()
    return s / s.sum()

def weighted_coverage(max_sim: np.ndarray, weights: np.ndarray, tau: float) -> float:
    return float((weights * (max_sim >= tau)).sum() / weights.sum())

def weighted_avgmax(max_sim: np.ndarray, weights: np.ndarray) -> float:
    return float((weights * max_sim).sum() / weights.sum())

def safe_f1(p: float, r: float) -> float:
    return 0.0 if (p + r) == 0 else float(2 * p * r / (p + r))

def describe_maxsim(name, arr):
    s = pd.Series(arr)
    desc = s.describe(percentiles=[.01, .05, .1, .25, .5, .75, .9, .95, .99])
    print(f"\n[{name} maxSim distribution]")
    display(desc.to_frame("value"))
    print(f"min={float(s.min()):.6f}, max={float(s.max()):.6f}")

# =============================
# 1) Load checklists
# =============================
A = load_checklist(path_a)
B = load_checklist(path_b)

print(f"Checklist A: {path_a} |A|={len(A)}")
print(f"Checklist B: {path_b} |B|={len(B)}")
display(A.head(5))
display(B.head(5))

# =============================
# 2) Category-level similarity (JSD/L1)
# =============================
pA = category_distribution(A)
pB = category_distribution(B)

cats = sorted(set(pA.index) | set(pB.index))
vecA = np.array([pA.get(c, 0.0) for c in cats], dtype=float)
vecB = np.array([pB.get(c, 0.0) for c in cats], dtype=float)

JSD = float((jensenshannon(vecA, vecB, base=2.0) ** 2))
L1 = float(np.abs(vecA - vecB).sum())

cat_df = pd.DataFrame({"category": cats, "A_weight_ratio": vecA, "B_weight_ratio": vecB})
print(f"\n[Category-level] JSD={JSD:.6f}, L1={L1:.6f}")
display(cat_df)

# 類別分佈：通常不是最重要
save_csv(cat_df, f"{OUT_PREFIX}_category_distribution.csv", enabled=(not IMPORTANT_ONLY))

plt.figure(figsize=(8, 4))
x = np.arange(len(cats))
plt.bar(x - 0.2, vecA, width=0.4, label="A")
plt.bar(x + 0.2, vecB, width=0.4, label="B")
plt.xticks(x, cats, rotation=30, ha="right")
plt.ylabel("weight ratio")
plt.title("Category weight distribution (A vs B)")
plt.legend()
plt.tight_layout()
if SAVE_FIG:
    save_fig(f"{OUT_PREFIX}_category_weight_distribution.png", enabled=(not IMPORTANT_ONLY))
plt.show()

# =============================
# 3) Embeddings + similarity matrix
# =============================
print("\nLoading embedding model:", EMB_MODEL)
model = SentenceTransformer(EMB_MODEL, trust_remote_code=True)

texts_A = A["text"].tolist()
texts_B = B["text"].tolist()

embA = model.encode(
    texts_A,
    normalize_embeddings=True,
    prompt_name=PROMPT_A,
    batch_size=ENCODE_BATCH_SIZE,
    show_progress_bar=True
)
embB = model.encode(
    texts_B,
    normalize_embeddings=True,
    prompt_name=PROMPT_B,
    batch_size=ENCODE_BATCH_SIZE,
    show_progress_bar=True
)

S = embA @ embB.T
print("Similarity matrix shape:", S.shape)

# =============================
# 4) Many-to-many coverage metrics
# =============================
wA = A["weight"].values.astype(float)
wB = B["weight"].values.astype(float)

maxA_to_B = S.max(axis=1)
bestB_for_A = S.argmax(axis=1)

maxB_to_A = S.max(axis=0)
bestA_for_B = S.argmax(axis=0)

avgmax_A_B = weighted_avgmax(maxA_to_B, wA)
avgmax_B_A = weighted_avgmax(maxB_to_A, wB)

print("\n[AvgMaxSim (no threshold)]")
print(f"AvgMaxSim_A→B = {avgmax_A_B:.6f}")
print(f"AvgMaxSim_B→A = {avgmax_B_A:.6f}")

# 先看 maxSim 分布：理解曲線會在哪裡開始掉
describe_maxsim("A→B", maxA_to_B)
describe_maxsim("B→A", maxB_to_A)

# 匹配集中度（很多 A 擠到少數 B 的話 coverage 會「看起來完美」）
vc = pd.Series(bestB_for_A).value_counts()
print("\n[Match concentration: A items -> B items]")
print(f"Unique matched B = {vc.size}/{len(B)}")
print(f"Top-1 B takes {int(vc.iloc[0])} A items (ratio={float(vc.iloc[0]/len(A)):.3f})")
print(f"Top-10 B takes ratio={float(vc.head(10).sum()/len(A)):.3f}")

# ✅ focus tau：用較高分位數，uncovered 才不會永遠 0
all_max = np.concatenate([maxA_to_B, maxB_to_A])
if AUTO_TAU_FOCUS:
    TAU_FOCUS = float(np.quantile(all_max, TAU_FOCUS_Q))

# ✅ 曲線/CSV 用固定 tau 網格（只算 0.5 -> 1.0）
TAUS = np.arange(PLOT_TAU_MIN, PLOT_TAU_MAX + 1e-12, PLOT_TAU_STEP)
TAUS = [float(round(x, 4)) for x in TAUS]

print("\n[TAU settings]")
print("TAU_FOCUS =", TAU_FOCUS)
print("TAUS head/tail:", TAUS[:3], "...", TAUS[-3:])

rows = []
for tau in TAUS:
    cov_A_B = weighted_coverage(maxA_to_B, wA, tau)  # recall-like
    cov_B_A = weighted_coverage(maxB_to_A, wB, tau)  # precision-like
    f1w = safe_f1(cov_B_A, cov_A_B)

    covA_cnt = float((maxA_to_B >= tau).mean())
    covB_cnt = float((maxB_to_A >= tau).mean())

    rows.append({
        "tau": tau,
        "lenA": len(A),
        "lenB": len(B),
        "Cov_A_to_B_weighted (recall-like)": cov_A_B,
        "Cov_B_to_A_weighted (precision-like)": cov_B_A,
        "F1_weighted": f1w,
        "Cov_A_to_B_count": covA_cnt,
        "Cov_B_to_A_count": covB_cnt,
        "AvgMaxSim_A_to_B": avgmax_A_B,
        "AvgMaxSim_B_to_A": avgmax_B_A
    })

m2m_df = pd.DataFrame(rows)
print("\n[Many-to-many metrics vs tau]")
display(m2m_df)

# ✅ 重要：永遠存
save_csv(m2m_df, f"{OUT_PREFIX}_m2m_metrics.csv", enabled=True)

# =============================
# Plot: single clean view (0.5 -> 1.0)
# =============================
tau = m2m_df["tau"].to_numpy()
y_ab = m2m_df["Cov_A_to_B_weighted (recall-like)"].to_numpy()
y_ba = m2m_df["Cov_B_to_A_weighted (precision-like)"].to_numpy()
y_f1 = m2m_df["F1_weighted"].to_numpy()

# marker 每隔幾個點才畫一次（step=0.01 時，5=每 0.05 一個 marker）
markevery = max(1, int(round(0.05 / PLOT_TAU_STEP)))

plt.figure(figsize=(9, 4.8))
plt.plot(tau, y_ab, label="Cov A→B (weighted)", marker="o",
         markevery=markevery, linewidth=2, markersize=4)
plt.plot(tau, y_ba, label="Cov B→A (weighted)", marker="o",
         markevery=markevery, linewidth=2, markersize=4)
plt.plot(tau, y_f1, label="F1 (weighted)", marker="o",
         markevery=markevery, linewidth=2, markersize=4)

plt.xlabel("tau (similarity threshold)")
plt.ylabel("score")
plt.title(f"Many-to-many rubric coverage vs tau (Nomic v2, {PROMPT_A}/{PROMPT_B})")

ax = plt.gca()
ax.set_xlim(0.5, 1.0)
ax.set_ylim(-0.02, 1.02)

ax.xaxis.set_major_locator(MultipleLocator(X_MAJOR_TICK))
ax.xaxis.set_minor_locator(MultipleLocator(X_MINOR_TICK))
ax.xaxis.set_major_formatter(FormatStrFormatter('%.2f'))

# 格線不要太吵：major 就好（需要 minor 再打開下一行）
ax.grid(True, which="major", alpha=0.25)
# ax.grid(True, which="minor", alpha=0.12)

ax.legend(loc="lower left", frameon=True)
plt.tight_layout()

FIG_NAME = f"{OUT_PREFIX}_m2m_tau_curve_0p5_to_1p0.png"
if SAVE_FIG:
    save_fig(FIG_NAME, enabled=True)

plt.show()

# =============================
# 5) Best-match tables (通常很大：IMPORTANT_ONLY=True 時不存)
# =============================
A_best = pd.DataFrame({
    "A_idx": np.arange(len(A)),
    "A_id": A["id"].values,
    "A_text": A["text"].values,
    "A_cat": A["category"].values,
    "A_w": wA,
    "best_B_idx": bestB_for_A,
    "best_B_id": B.loc[bestB_for_A, "id"].values,
    "best_B_text": B.loc[bestB_for_A, "text"].values,
    "best_B_cat": B.loc[bestB_for_A, "category"].values,
    "best_B_w": B.loc[bestB_for_A, "weight"].values,
    "maxSim": maxA_to_B
}).sort_values("maxSim", ascending=False)

B_best = pd.DataFrame({
    "B_idx": np.arange(len(B)),
    "B_id": B["id"].values,
    "B_text": B["text"].values,
    "B_cat": B["category"].values,
    "B_w": wB,
    "best_A_idx": bestA_for_B,
    "best_A_id": A.loc[bestA_for_B, "id"].values,
    "best_A_text": A.loc[bestA_for_B, "text"].values,
    "best_A_cat": A.loc[bestA_for_B, "category"].values,
    "best_A_w": A.loc[bestA_for_B, "weight"].values,
    "maxSim": maxB_to_A
}).sort_values("maxSim", ascending=False)

display(A_best.head(10))
display(B_best.head(10))

save_csv(A_best, f"{OUT_PREFIX}_A_to_B_bestmatch.csv", enabled=(not IMPORTANT_ONLY))
save_csv(B_best, f"{OUT_PREFIX}_B_to_A_bestmatch.csv", enabled=(not IMPORTANT_ONLY))

# =============================
# 6) Uncovered items at tau_focus（重要：差異清單）
# =============================
A_uncovered = A_best[A_best["maxSim"] < TAU_FOCUS].copy()
B_uncovered = B_best[B_best["maxSim"] < TAU_FOCUS].copy()

print(f"\n[Uncovered items at tau={TAU_FOCUS:.4f}]")
print(f"A uncovered: {len(A_uncovered)}/{len(A)}")
print(f"B uncovered: {len(B_uncovered)}/{len(B)}")

display(A_uncovered)
display(B_uncovered)

# ✅ 重要：永遠存
save_csv(A_uncovered, f"{OUT_PREFIX}_A_uncovered_tau{TAU_FOCUS:.4f}.csv", enabled=True)
save_csv(B_uncovered, f"{OUT_PREFIX}_B_uncovered_tau{TAU_FOCUS:.4f}.csv", enabled=True)

print("\nSaved outputs to:", OUT_DIR.resolve())
print("Important outputs:")
print(f" - {OUT_PREFIX}_m2m_metrics.csv")
if SAVE_FIG:
    print(f" - {FIG_NAME}")
print(f" - {OUT_PREFIX}_A_uncovered_tau{TAU_FOCUS:.4f}.csv")
print(f" - {OUT_PREFIX}_B_uncovered_tau{TAU_FOCUS:.4f}.csv")
```

# 整體評分趨勢一致性

```python
import json
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

# =============================
# Config（改成你電腦的相對路徑）
# =============================
judge_a_path = Path("Qwen3_Instruct") / "phase3_scores_Qwen3VL32B.jsonl"
judge_b_path = Path("Gemma3") / "phase3_scores_Gemma3.jsonl"

OUT_DIR = Path("outputs")
OUT_DIR.mkdir(parents=True, exist_ok=True)

OUT_PREFIX = "phase3_score_agreement_Qwen_vs_Gemma"

# 相關性 bootstrap
BOOT_CORR = 5000        # 想更穩可 10000（會更慢）
SEED = 123

# 固定門檻一致性
THRESHOLDS = [0.8, 0.9, 1.0]

# low-quality（bottom-p%）分析
P_LIST = [0.05, 0.10, 0.20]              # 針對幾個 p 直接輸出清單
P_GRID = np.arange(0.01, 0.301, 0.01)    # 敏感度曲線 p=1%..30%
SAVE_LOW_ID_LISTS = True                 # 會存 Aonly/Bonly/Both 的 id 清單

# 分箱一致性（QWK）
N_BINS = 10

SAVE_FIG = True


# =============================
# Helpers
# =============================
def outpath(name: str) -> Path:
    return OUT_DIR / name

def save_csv(df: pd.DataFrame, filename, enabled: bool = True, index: bool = False, encoding: str = "utf-8-sig"):
    """
    filename 可以是 str 或 Path
    - 若是 str：會自動存到 OUT_DIR/name.csv
    - 若是 Path：直接用該 Path
    """
    if not enabled:
        return
    fpath = filename if isinstance(filename, Path) else outpath(filename)
    df.to_csv(fpath, index=index, encoding=encoding)

def load_scores_jsonl(jsonl_path: Path):
    """回傳 dict[id->score], n_lines, missing_score_lines"""
    if not jsonl_path.exists():
        raise FileNotFoundError(f"找不到檔案：{jsonl_path.resolve()}")

    scores = {}
    n_lines = 0
    missing = 0
    with open(jsonl_path, "r", encoding="utf-8") as f:
        for line in f:
            n_lines += 1
            o = json.loads(line)
            sid = str(o.get("id"))
            sc = o.get("score", None)
            if sc is None:
                missing += 1
                continue
            scores[sid] = float(sc)
    return scores, n_lines, missing

def bootstrap_ci_corr(x, y, corr_func, n_boot=2000, seed=123):
    rng = np.random.default_rng(seed)
    n = len(x)
    idx = np.arange(n)
    vals = np.empty(n_boot, dtype=np.float64)
    for b in range(n_boot):
        samp = rng.choice(idx, size=n, replace=True)
        vals[b] = corr_func(x[samp], y[samp])
    lo, hi = np.quantile(vals, [0.025, 0.975])
    return float(lo), float(hi)

def bottom_set(x, ids, p):
    k = int(np.ceil(len(x) * p))
    idx = np.argsort(x)[:k]  # lowest scores
    return set(ids[i] for i in idx)

def set_metrics(LA, LB):
    inter = len(LA & LB)
    union = len(LA | LB)
    jacc = inter / union if union else np.nan
    overlap = inter / min(len(LA), len(LB)) if min(len(LA), len(LB)) else np.nan
    precision = inter / len(LA) if len(LA) else np.nan
    recall = inter / len(LB) if len(LB) else np.nan
    f1 = (2*precision*recall/(precision+recall)) if (precision+recall) else np.nan
    return inter, jacc, overlap, precision, recall, f1

def quantile_bins(x, n_bins=10):
    qs = np.quantile(x, np.linspace(0, 1, n_bins + 1))
    return np.digitize(x, qs[1:-1], right=True)  # 0..n_bins-1

def confusion_matrix(a, b, n):
    cm = np.zeros((n, n), dtype=np.int64)
    for i in range(len(a)):
        cm[a[i], b[i]] += 1
    return cm

def quadratic_weighted_kappa(a, b, n):
    cm = confusion_matrix(a, b, n).astype(np.float64)
    N = cm.sum()
    if N == 0:
        return np.nan, cm

    W = np.zeros((n, n), dtype=np.float64)
    for i in range(n):
        for j in range(n):
            W[i, j] = ((i - j) ** 2) / ((n - 1) ** 2)

    row = cm.sum(axis=1)
    col = cm.sum(axis=0)
    E = np.outer(row, col) / N

    num = (W * cm).sum()
    den = (W * E).sum()
    kappa = 1 - num / den if den > 0 else np.nan
    return float(kappa), cm

def discrimination_diag(x, name):
    q1, q3 = np.quantile(x, [0.25, 0.75])
    iqr = q3 - q1
    lt08 = (x < 0.8).mean()
    lt09 = (x < 0.9).mean()

    xr3 = np.round(x, 3)
    xr4 = np.round(x, 4)
    ur3 = len(np.unique(xr3)) / len(xr3)
    ur4 = len(np.unique(xr4)) / len(xr4)

    _, counts = np.unique(xr3, return_counts=True)
    ties_proxy = counts.max() / len(x)

    return {
        "judge": name,
        "mean": float(x.mean()),
        "std": float(x.std(ddof=1)),
        "iqr": float(iqr),
        "P(score<0.8)": float(lt08),
        "P(score<0.9)": float(lt09),
        "unique_ratio@3dp": float(ur3),
        "unique_ratio@4dp": float(ur4),
        "ties_proxy@3dp(maxfreq/N)": float(ties_proxy),
    }


# =============================
# 1) Load + align
# =============================
A, nA, missA = load_scores_jsonl(judge_a_path)
B, nB, missB = load_scores_jsonl(judge_b_path)

ids = sorted(set(A.keys()) & set(B.keys()))
if len(ids) == 0:
    raise RuntimeError("兩個檔案沒有重疊的 id（intersection=0）")

xa = np.array([A[i] for i in ids], dtype=np.float64)
xb = np.array([B[i] for i in ids], dtype=np.float64)

print(f"Judge A file: {judge_a_path} | lines={nA}, missing_score_lines={missA}, unique_ids={len(A)}")
print(f"Judge B file: {judge_b_path} | lines={nB}, missing_score_lines={missB}, unique_ids={len(B)}")
print(f"Intersection N={len(ids)}")
print(f"A mean/std = {xa.mean():.6f} / {xa.std(ddof=1):.6f}")
print(f"B mean/std = {xb.mean():.6f} / {xb.std(ddof=1):.6f}")

diag_df = pd.DataFrame([
    discrimination_diag(xa, "A(Qwen)"),
    discrimination_diag(xb, "B(Gemma)")
])
print("\n[Discrimination diagnostics]")
display(diag_df)
save_csv(diag_df, f"{OUT_PREFIX}_discrimination_diag.csv", enabled=True)

if SAVE_FIG:
    plt.figure()
    plt.hist(xa, bins=50, alpha=0.6, label="Judge A (Qwen)")
    plt.hist(xb, bins=50, alpha=0.6, label="Judge B (Gemma)")
    plt.xlabel("score")
    plt.ylabel("count")
    plt.title("Score distribution")
    plt.legend()
    plt.tight_layout()
    plt.savefig(outpath(f"{OUT_PREFIX}_score_hist.png"), dpi=200)
    plt.show()

# =============================
# 2) Trend consistency: correlations + p-values
# =============================
r, p_r = stats.pearsonr(xa, xb)
rho, p_rho = stats.spearmanr(xa, xb)
tau, p_tau = stats.kendalltau(xa, xb, variant="b")

print("\n[Trend consistency]")
print(f"Pearson r      = {r:.6f} (p={p_r:.3e})")
print(f"Spearman rho   = {rho:.6f} (p={p_rho:.3e})")
print(f"Kendall tau-b  = {tau:.6f} (p={p_tau:.3e})")

pearson_func = lambda x, y: stats.pearsonr(x, y)[0]
spearman_fast = lambda x, y: stats.pearsonr(stats.rankdata(x), stats.rankdata(y))[0]

ci_r = bootstrap_ci_corr(xa, xb, pearson_func, n_boot=BOOT_CORR, seed=SEED)
ci_rho = bootstrap_ci_corr(xa, xb, spearman_fast, n_boot=BOOT_CORR, seed=SEED)

print("\n[Bootstrap 95% CI]")
print(f"Pearson r CI    = ({ci_r[0]:.6f}, {ci_r[1]:.6f})  [boot={BOOT_CORR}]")
print(f"Spearman rho CI = ({ci_rho[0]:.6f}, {ci_rho[1]:.6f})  [boot={BOOT_CORR}]")

corr_summary = pd.DataFrame([{
    "N": len(ids),
    "Pearson_r": r, "Pearson_p": p_r, "Pearson_CI_lo": ci_r[0], "Pearson_CI_hi": ci_r[1],
    "Spearman_rho": rho, "Spearman_p": p_rho, "Spearman_CI_lo": ci_rho[0], "Spearman_CI_hi": ci_rho[1],
    "Kendall_tau_b": tau, "Kendall_p": p_tau,
}])
save_csv(corr_summary, f"{OUT_PREFIX}_correlation_summary.csv", enabled=True)

# =============================
# 3) Threshold agreement (score >= t)
# =============================
print("\n[Threshold agreement]")
th_rows = []
for t in THRESHOLDS:
    pa = xa >= t
    pb = xb >= t
    agree = (pa == pb).mean()
    th_rows.append({
        "threshold": t,
        "agreement": float(agree),
        "pass_rate_A": float(pa.mean()),
        "pass_rate_B": float(pb.mean())
    })
    print(f"t={t:.2f} | agreement={agree:.6f} | pass_rate(A)={pa.mean():.6f} | pass_rate(B)={pb.mean():.6f}")

th_df = pd.DataFrame(th_rows)
save_csv(th_df, f"{OUT_PREFIX}_threshold_agreement.csv", enabled=True)

# =============================
# 4) Bottom-p% overlap (low-quality agreement)
# =============================
print("\n[Bottom-p% overlap (low-quality set agreement)]")
rows = []
N = len(ids)

for p in P_LIST:
    LA = bottom_set(xa, ids, p)
    LB = bottom_set(xb, ids, p)
    inter, jacc, overlap, prec, rec, f1 = set_metrics(LA, LB)
    exp_inter = (p * p) * N
    lift = inter / exp_inter if exp_inter > 0 else np.nan

    rows.append({
        "p(bottom)": p,
        "|LA|": len(LA),
        "|LB|": len(LB),
        "intersection": inter,
        "expected_random(p^2*N)": exp_inter,
        "lift_over_random": float(lift),
        "Jaccard": float(jacc),
        "OverlapCoeff": float(overlap),
        "Precision(A->B)": float(prec),
        "Recall(A->B)": float(rec),
        "F1(A vs B)": float(f1),
    })

    print(f"p={p:.2f} | inter={inter} | random~{exp_inter:.1f} | lift={lift:.3f} | Jacc={jacc:.4f} | F1={f1:.4f}")

bottom_df = pd.DataFrame(rows)
display(bottom_df)
save_csv(bottom_df, f"{OUT_PREFIX}_bottomp_summary.csv", enabled=True)

def save_low_id_lists(p):
    LA = bottom_set(xa, ids, p)
    LB = bottom_set(xb, ids, p)
    both = sorted(list(LA & LB))
    a_only = sorted(list(LA - LB))
    b_only = sorted(list(LB - LA))
    save_csv(pd.DataFrame({"id": both}), f"{OUT_PREFIX}_low_both_p{int(p*100)}.csv", enabled=True)
    save_csv(pd.DataFrame({"id": a_only}), f"{OUT_PREFIX}_low_Aonly_p{int(p*100)}.csv", enabled=True)
    save_csv(pd.DataFrame({"id": b_only}), f"{OUT_PREFIX}_low_Bonly_p{int(p*100)}.csv", enabled=True)

if SAVE_LOW_ID_LISTS:
    for p in P_LIST:
        save_low_id_lists(p)
    print("\nSaved low-quality ID lists to outputs/ (both/Aonly/Bonly).")

# =============================
# 5) Sensitivity curve for bottom-p% (p=1%..30%)
# =============================
# =============================
# 5) Sensitivity curve for bottom-p% (with 95% bootstrap band)
# =============================
P_GRID = np.arange(0.01, 0.301, 0.01)  # 1%..30%

N_BOOT = 500        # 你現在圖上是 B=500；想更穩可 2000/5000（會更慢）
SEED_BAND = 123
CI_LO, CI_HI = 0.025, 0.975

def sensitivity_point_estimate(xa, xb, p_grid):
    """
    在原始資料上計算 bottom-p% 的 Jaccard/F1/Lift 曲線（點估計）
    用「增量 grow bottom-k」寫法，比每個 p 用 set 重算快很多
    """
    N = len(xa)
    order_a = np.argsort(xa)   # low -> high
    order_b = np.argsort(xb)

    mask_a = np.zeros(N, dtype=bool)
    mask_b = np.zeros(N, dtype=bool)
    inter = 0
    prev_k = 0

    jacc = np.zeros(len(p_grid), dtype=np.float64)
    f1   = np.zeros(len(p_grid), dtype=np.float64)
    lift = np.zeros(len(p_grid), dtype=np.float64)

    for t, p in enumerate(p_grid):
        k = int(np.ceil(N * p))
        if k <= prev_k:
            k = prev_k + 1

        newA = order_a[prev_k:k]
        inter += np.count_nonzero(mask_b[newA])
        mask_a[newA] = True

        newB = order_b[prev_k:k]
        # 注意：避免 double count，只加上「新加入 B」且已在 A 的
        inter += np.count_nonzero(mask_a[newB] & (~mask_b[newB]))
        mask_b[newB] = True

        union = 2 * k - inter
        jacc[t] = inter / union if union > 0 else np.nan

        # bottom-k vs bottom-k：precision=recall=inter/k => F1 = inter/k
        f1[t] = inter / k if k > 0 else np.nan

        exp_inter = (p * p) * N
        lift[t] = inter / exp_inter if exp_inter > 0 else np.nan

        prev_k = k

    return jacc, f1, lift

def bootstrap_bands(xa, xb, p_grid, n_boot=500, seed=123, qlo=0.025, qhi=0.975):
    """
    bootstrap 95% band：
    抽 N 筆有放回 -> 計算整條曲線 -> 對每個 p 取分位數形成 band
    """
    rng = np.random.default_rng(seed)
    N = len(xa)
    P = len(p_grid)

    jacc_mat = np.zeros((n_boot, P), dtype=np.float64)
    f1_mat   = np.zeros((n_boot, P), dtype=np.float64)
    lift_mat = np.zeros((n_boot, P), dtype=np.float64)

    idx_all = np.arange(N)

    for b in range(n_boot):
        samp = rng.choice(idx_all, size=N, replace=True)
        xa_s = xa[samp]
        xb_s = xb[samp]

        j, f, l = sensitivity_point_estimate(xa_s, xb_s, p_grid)
        jacc_mat[b] = j
        f1_mat[b]   = f
        lift_mat[b] = l

    j_lo = np.quantile(jacc_mat, qlo, axis=0)
    j_hi = np.quantile(jacc_mat, qhi, axis=0)
    f_lo = np.quantile(f1_mat,   qlo, axis=0)
    f_hi = np.quantile(f1_mat,   qhi, axis=0)
    l_lo = np.quantile(lift_mat, qlo, axis=0)
    l_hi = np.quantile(lift_mat, qhi, axis=0)

    return (j_lo, j_hi), (f_lo, f_hi), (l_lo, l_hi)

# 1) 點估計曲線
j_point, f_point, l_point = sensitivity_point_estimate(xa, xb, P_GRID)

# 2) bootstrap bands
(j_lo, j_hi), (f_lo, f_hi), (l_lo, l_hi) = bootstrap_bands(
    xa, xb, P_GRID, n_boot=N_BOOT, seed=SEED_BAND, qlo=CI_LO, qhi=CI_HI
)

# 3) 存成 CSV（含 point + band）
sens_band_df = pd.DataFrame({
    "p": P_GRID,
    "jaccard": j_point,
    "jacc_lo": j_lo,
    "jacc_hi": j_hi,
    "f1": f_point,
    "f1_lo": f_lo,
    "f1_hi": f_hi,
    "lift": l_point,
    "lift_lo": l_lo,
    "lift_hi": l_hi
})
save_csv(sens_band_df, f"{OUT_PREFIX}_bottom_p_sensitivity_band.csv", enabled=True)

# 4) 圖：Jaccard & F1（含 band）
plt.figure()
plt.plot(P_GRID, j_point, label="Jaccard (bottom-p%)")
plt.fill_between(P_GRID, j_lo, j_hi, alpha=0.25)
plt.plot(P_GRID, f_point, label="F1 (bottom-p%)")
plt.fill_between(P_GRID, f_lo, f_hi, alpha=0.25)
plt.xlabel("p (bottom proportion)")
plt.ylabel("overlap score")
plt.title(f"Sensitivity curve (bottom-p%) with 95% bootstrap band (B={N_BOOT})")
plt.legend()
plt.tight_layout()
if SAVE_FIG:
    plt.savefig(outpath(f"{OUT_PREFIX}_bottomp_sensitivity_overlap_band.png"), dpi=200)
plt.show()

# 5) 圖：Lift over random（含 band）
plt.figure()
plt.plot(P_GRID, l_point, label="Lift over random")
plt.fill_between(P_GRID, l_lo, l_hi, alpha=0.25)
plt.axhline(1.0, linestyle="--")
plt.xlabel("p (bottom proportion)")
plt.ylabel("lift over random")
plt.title(f"Lift over random baseline with 95% bootstrap band (B={N_BOOT})")
plt.legend()
plt.tight_layout()
if SAVE_FIG:
    plt.savefig(outpath(f"{OUT_PREFIX}_bottomp_sensitivity_lift_band.png"), dpi=200)
plt.show()

print("Saved (band plots & csv):")
print(" -", outpath(f"{OUT_PREFIX}_bottomp_sensitivity_overlap_band.png").resolve())
print(" -", outpath(f"{OUT_PREFIX}_bottomp_sensitivity_lift_band.png").resolve())
print(" -", outpath(f"{OUT_PREFIX}_bottom_p_sensitivity_band.csv").resolve())

# =============================
# 6) Quantile binning agreement + QWK
# =============================
ba = quantile_bins(xa, n_bins=N_BINS)
bb = quantile_bins(xb, n_bins=N_BINS)
qwk, cm = quadratic_weighted_kappa(ba, bb, N_BINS)

print("\n[Binning agreement]")
print(f"Quantile bins={N_BINS}, Quadratic Weighted Kappa (QWK)={qwk:.6f}")

if SAVE_FIG:
    plt.figure()
    plt.imshow(cm, aspect="auto")
    plt.colorbar()
    plt.title("Confusion matrix of quantile bins (A rows, B cols)")
    plt.xlabel("B bin")
    plt.ylabel("A bin")
    plt.tight_layout()
    plt.savefig(outpath(f"{OUT_PREFIX}_quantile_bins_confusion.png"), dpi=200)
    plt.show()

# =============================
# 7) Export one-line summary for reporting
# =============================
summary = {
    "N_intersection": N,
    "A_mean": float(xa.mean()),
    "A_std": float(xa.std(ddof=1)),
    "B_mean": float(xb.mean()),
    "B_std": float(xb.std(ddof=1)),
    "Pearson_r": float(r),
    "Pearson_p": float(p_r),
    "Pearson_CI_lo": float(ci_r[0]),
    "Pearson_CI_hi": float(ci_r[1]),
    "Spearman_rho": float(rho),
    "Spearman_p": float(p_rho),
    "Spearman_CI_lo": float(ci_rho[0]),
    "Spearman_CI_hi": float(ci_rho[1]),
    "Kendall_tau_b": float(tau),
    "Kendall_p": float(p_tau),
    "QWK_bins": float(qwk),
}
save_csv(pd.DataFrame([summary]), f"{OUT_PREFIX}_summary.csv", enabled=True)

print("\nSaved summary:", outpath(f"{OUT_PREFIX}_summary.csv").resolve())
print("Saved sensitivity:", outpath(f"{OUT_PREFIX}_bottom_p_sensitivity.csv").resolve())
print("All outputs in:", OUT_DIR.resolve())
```
